In NLP competitions, we frequently need to perform inference in a notebook that does not have internet access. This means that we need to instatiate transformer models offline.

This notebook demonstartes how to use the [pretrained transformers dataset](https://www.kaggle.com/kozodoi/transformers) to initialize some of the most popular `huggingface` transfomers models from the locally saved pre-trained weights, tokenizers and configuration files.

First, we need to import the `transformers` library.

```python
##### INSTALLATION

import transformers
from transformers import AutoModel, AutoTokenizer
transformers.__version__
```

The `transformers` package works with many NLP models. See [model hub](https://huggingface.co/models) with the full list of models. 

The [pretrained transformers dataset](https://www.kaggle.com/kozodoi/transformers) dataset currently includes the following models:
- `albert-large-v2`
- `bert-base-uncased`
- `bert-large-uncased`
- `distilroberta-base`
- `distilbert-base-uncased`
- `google/electra-base-discriminator`
- `facebook/bart-base`
- `facebook/bart-large`
- `funnel-transformer/small`
- `funnel-transformer/large`
- `roberta-base`
- `roberta-large`
- `t5-base`
- `t5-large`
- `xlnet-base-cased`
- `xlnet-large-cased`

You can use the dataset as follows:
- When Internet access is enabled (in the training notebooks), we can simply specify the model name in the model initialization function.
- When Internet access is disabled (in the inference notebooks), we can instantiate a model from the pre-trained weights and confidgurations saved in the [pretrained transformers dataset](https://www.kaggle.com/kozodoi/transformers).

Let's illustrate this by instantiating a Roberta model

```python
##### INSTANTIATE MODEL

# model name (if Internet is enabled)
# model path (if Internet is disabled)
model_path_or_name = '../input/transformers/roberta-base'

# instantiate model & tokenizer
model     = AutoModel.from_pretrained(model_path_or_name)
tokenizer = AutoTokenizer.from_pretrained(model_path_or_name)
```
